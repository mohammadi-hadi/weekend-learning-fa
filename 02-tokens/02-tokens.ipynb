## ۱. هر حرف فارسی دو بایت است

کامپیوتر متن را به صورت بایت (byte) ذخیره می‌کند. در یوتی‌اف ۸ (UTF-8) هر حرف انگلیسی یک بایت است و هر حرف فارسی دو بایت. کد زیر همین را برای جمله‌ی درس و چند حرف نشان می‌دهد.

</div>

```python
fa = "فارسی دستور زبانی نسبتاً آسان و عمدتاً منظم دارد."
en = "Persian has a relatively easy and mostly regular grammar."

for name, text in (("فارسی", fa), ("انگلیسی", en)):
    print(name, "| حرف:", len(text), "| بایت:", len(text.encode("utf-8")))

print()
for ch in ("ز", "ب", "a", "z"):
    print(ch, "->", ch.encode("utf-8"), len(ch.encode("utf-8")), "بایت")
```

## ۲. تکه‌های هر توکن‌ساز

کد زیر تکه‌ها را از فایل `data.json` کنار همین دفترچه می‌خواند؛ این تکه‌ها یک بار با کتابخانه‌ی توکن‌سازهای چت‌جی‌پی‌تی حساب شده‌اند. تکه‌ها با خط عمودی از هم جدا شده‌اند. تکه‌ای که با فاصله شروع می‌شود اول یک کلمه‌ی تازه است، و تکه‌ای مثل `<0xD8>` فقط یک بایت است، یعنی نیمی از یک حرف فارسی.

</div>

```python
import json
from pathlib import Path

path = Path("data.json")
if not path.exists():
    raise SystemExit("data.json کنار این دفترچه نیست؛ دفترچه را از پوشه‌ی درس اجرا کنید.")
data = json.loads(path.read_text(encoding="utf-8"))

for text in data["texts"]:
    print("==", text["label"])
    for tok in data["tokenizers"]:
        pieces = text["by"][tok["id"]]
        shown = " | ".join(p.replace(" ", "_") for p in pieces)
        print(f"  {tok['label']}: {len(pieces)} توکن")
        print("   ", shown)
    print()
```

<div dir="rtl">

## ۳. در کل دو هزار جمله

همان جمله‌ها به دو زبان، یک بار با هر توکن‌ساز شمرده شده‌اند. عدد آخر یعنی برای هر توکن انگلیسی چند توکن فارسی لازم شد.

</div>

```python
for tok in data["tokenizers"]:
    c = data["corpus"][tok["id"]]
    print(f"{tok['label']}: انگلیسی {c['en_tokens']:,} | فارسی {c['fa_tokens']:,} | نسبت {c['ratio']:.2f}")
```

## ۴. اگر کتابخانه‌ی tiktoken را دارید (اختیاری)

این سلول فقط وقتی کار می‌کند که کتابخانه‌ی `tiktoken` نصب باشد؛ نصبش اینترنت می‌خواهد. اگر نصب نیست، فقط پیامی چاپ می‌کند و بقیه‌ی دفترچه همچنان درست است. با آن می‌توانید جمله‌های خودتان را هم بشمارید.

</div>

```python
try:
    import tiktoken
except ImportError:
    print("tiktoken نصب نیست؛ این سلول اختیاری است و می‌توانید از آن بگذرید.")
else:
    mine = "یک جمله‌ی فارسی خودتان را اینجا بنویسید."
    for name in ("cl100k_base", "o200k_base"):
        enc = tiktoken.get_encoding(name)
        print(name, len(enc.encode(mine)), "توکن")
```
