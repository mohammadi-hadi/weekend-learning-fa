## ۱. شش سفر و خطای یک حدس

کد زیر سفرها را از فایل `data.json` کنار همین دفترچه می‌خواند. مسافت به کیلومتر است و کرایه به هزار تومان. بعد برای یک حدس، کرایه‌ی پایه ۲۰ و نرخ ۱۰، کرایه‌ی هر سفر را پیش‌بینی می‌کند و خطایش را نشان می‌دهد.

</div>

```python
import json
from pathlib import Path

path = Path("data.json")
if not path.exists():
    raise SystemExit("data.json کنار این دفترچه نیست؛ دفترچه را از پوشه‌ی درس اجرا کنید.")
data = json.loads(path.read_text(encoding="utf-8"))
km = [t["km"] for t in data["trips"]]
fare = [t["fare"] for t in data["trips"]]


def errors(base, rate):
    return [base + rate * k - f for k, f in zip(km, fare)]


base, rate = 20, 10
for k, f, e in zip(km, fare, errors(base, rate)):
    print(f"{k:>2} کیلومتر | کرایه {f:>3} | پیش‌بینی {base + rate * k:>5} | خطا {e:+}")
print("جمع مربع خطاها:", sum(e * e for e in errors(base, rate)))
```

## ۲. بهترین دو عدد با امتحان کردن

کد زیر همه‌ی کرایه‌های پایه از ۰ تا ۶۰ و همه‌ی نرخ‌ها از ۰ تا ۱۶، با گام نیم، را امتحان می‌کند و جفتی را نگه می‌دارد که جمع مربع خطاهایش کمترین است. این کار کند است، ولی ایده همین است.

</div>

```python
best = None
for b in range(0, 61):
    for r2 in range(0, 33):
        r = r2 / 2
        sse = sum(e * e for e in errors(b, r))
        if best is None or sse < best[0]:
            best = (sse, b, r)
print("کمترین جمع مربع خطاها:", best[0], "| کرایه‌ی پایه:", best[1], "| نرخ:", best[2])
```

## ۳. همان جواب با کمترین مربعات

کمترین مربعات (least squares) همان جفت را بی‌آنکه چیزی را امتحان کند، با یک فرمول پیدا می‌کند. بعد کرایه‌ی سفر هفتم، ۷ کیلومتر، را پیش‌بینی می‌کند.

</div>

```python
n = len(km)
mk, mf = sum(km) / n, sum(fare) / n
rate = sum((k - mk) * (f - mf) for k, f in zip(km, fare)) / sum((k - mk) ** 2 for k in km)
base = mf - rate * mk
print(round(base, 2), round(rate, 2))
print(round(base + rate * data["seventh_km"], 2))
```
