# فارسی، زبانی کم‌منبع

این دفترچه‌ی ژوپیتر (Jupyter notebook) عددهای درس چهارم را با پایتون (Python) از داده‌های خود درس حساب می‌کند: از هر هزار صفحه‌ی یک جمع‌آوری کامن‌کرال (Common Crawl) چندتا به هر زبان است، سهم فارسی در ده جمع‌آوری چقدر بالا و پایین رفته و در داده‌های آموزشی جی‌پی‌تی ۳ از هر ده هزار کلمه چندتا فارسی بود. کدها جز خود پایتون چیزی نمی‌خواهند و اینترنت هم لازم ندارند؛ سلول‌ها را به ترتیب اجرا کنید.

سهم زبان‌ها از جدولی است که خود کامن‌کرال منتشر می‌کند. زبان هر صفحه را یک برنامه‌ی خودکار تشخیص داده و این برنامه ممکن است اشتباه کند.

## ۱. هزار صفحه از جمع‌آوری شهریور

کد زیر داده‌ها را از فایل `data.json` کنار همین دفترچه می‌خواند. بعد حساب می‌کند که از هر هزار صفحه‌ی جمع‌آوری شهریور ۱۴۰۵ حدود چندتا انگلیسی، هلندی و فارسی بود و در برابر هر صفحه‌ی فارسی چند صفحه به دو زبان دیگر بود.

</div>

```python
import json
from pathlib import Path

path = Path("data.json")
if not path.exists():
    raise SystemExit("data.json کنار این دفترچه نیست؛ دفترچه را از پوشه‌ی درس اجرا کنید.")
data = json.loads(path.read_text(encoding="utf-8"))
crawl = data["crawl"]
names = data["names"]
share = {row["code"]: row["pct"] for row in data["languages"]}

other = "nld"
print("جمع‌آوری", crawl["month"], crawl["year"], "|", crawl["id"])
for code in ["eng", other, "fas"]:
    print(names.get(code, code), "| حدود", round(share[code] * 10), "صفحه از هر هزار صفحه")
print("صفحه‌ی انگلیسی در برابر هر صفحه‌ی فارسی:", round(share["eng"] / share["fas"]))
print("صفحه‌ی", names.get(other, other), "در برابر هر صفحه‌ی فارسی:", round(share[other] / share["fas"], 1))
```

## ۲. جواب حساب‌های قلم و کاغذ

در بخش «ده دقیقه با قلم و کاغذ» سهم فارسی در جمع‌آوری شهریور ۰٫۶۹۳۴٪ بود و سه حساب از شما خواسته شد: چند صفحه از هر هزار صفحه، چندتا از هر یک میلیون صفحه و چندتا در کل آن جمع‌آوری. اول روی کاغذ حساب کنید. بعد در سلول زیر به جای `False` بنویسید `True` و سلول را دوباره اجرا کنید تا جواب‌ها را ببینید.

</div>

```python
show_answers = False

pct = share["fas"]
if show_answers:
    print("از هر هزار صفحه:", round(pct / 100 * 1000, 3))
    print("از هر یک میلیون صفحه:", round(pct / 100 * 1_000_000))
    print("در کل جمع‌آوری، با 2.2 میلیارد صفحه: حدود", round(pct / 100 * 2.2e9 / 1e6, 1), "میلیون صفحه")
    print("با تعداد دقیق صفحه‌ها،", crawl["total_pages"], "صفحه: حدود",
          round(pct / 100 * crawl["total_pages"] / 1e6, 2), "میلیون صفحه")
else:
    print("اول روی کاغذ حساب کنید، بعد show_answers را True کنید و این سلول را دوباره اجرا کنید.")
```

## ۳. سهم فارسی در ده جمع‌آوری

کد زیر برای هر جمع‌آوری سال ۲۰۲۶ نشان می‌دهد که از هر هزار صفحه حدود چندتا فارسی و چندتا هلندی بود؛ هر خانه‌ی پر یک صفحه است. هر جمع‌آوری حدود دو هفته طول کشید و به اسم ماهی آمده که بیشتر آن دو هفته در آن بود. دو خط آخر کمترین و بیشترین سهم را پیدا می‌کنند.

</div>

```python
for row in data["series"]:
    fa = round(row["fas_pct"] * 10)
    nl = round(row["nld_pct"] * 10)
    print(row["id"], "|", row["month"], row["year"])
    print("    فارسی", f"{fa:>2}", "█" * fa)
    print("    هلندی", f"{nl:>2}", "█" * nl)

key = "fas_pct"
label = {"fas_pct": "فارسی", "nld_pct": "هلندی", "eng_pct": "انگلیسی"}[key]
lo = min(data["series"], key=lambda row: row[key])
hi = max(data["series"], key=lambda row: row[key])
print("کمترین سهم", label + ":", lo["month"], lo["year"], "| حدود", round(lo[key] * 10), "از هر هزار")
print("بیشترین سهم", label + ":", hi["month"], hi["year"], "| حدود", round(hi[key] * 10), "از هر هزار")
```

## ۴. زبان‌هایی که از فارسی صفحه‌ی بیشتری داشتند

کد زیر زبان‌های جمع‌آوری شهریور را به ترتیب تعداد صفحه مرتب می‌کند و آن‌هایی را نشان می‌دهد که از فارسی صفحه‌ی بیشتری داشتند. صفحه‌هایی که زبانشان تشخیص داده نشده، جدا آمده‌اند. اگر اسم فارسی زبانی در داده‌ها نباشد، کد سه‌حرفی‌اش را می‌بینید.

</div>

```python
than = "fas"
known = [row for row in data["languages"] if row["code"] != "<unknown>"]
known.sort(key=lambda row: row["pct"], reverse=True)
above = [row for row in known if row["pct"] > share[than]]
for row in above:
    print(names.get(row["code"], row["code"]), "| حدود", round(row["pct"] * 10), "از هر هزار")
print(len(above), "زبان از", names.get(than, than), "صفحه‌ی بیشتری داشتند.")
print(names["<unknown>"], "| حدود", round(share["<unknown>"] * 10), "از هر هزار")
```

## ۵. کلمه‌های فارسی در داده‌های آموزشی جی‌پی‌تی ۳

اوپن‌ای‌آی همراه مقاله‌ی جی‌پی‌تی ۳ تعداد کلمه‌های هر زبان را در داده‌های آموزشی‌اش منتشر کرد. واحد اینجا کلمه است، نه صفحه؛ پس این عددها را کنار عددهای کامن‌کرال نگذارید.

</div>

```python
words = data["gpt3"]["words"]
total = data["gpt3"]["total_words"]
print("انگلیسی:", round(words["en"] / total * 100), "درصد کلمه‌ها")
for code, name in [("nl", "هلندی"), ("fa", "فارسی")]:
    print(name + ":", round(words[code] / total * 10_000, 2), "کلمه از هر ده هزار کلمه")
print("کلمه‌ی هلندی در برابر هر کلمه‌ی فارسی:", round(words["nl"] / words["fa"]))
```
